```python
import numpy as np
import matplotlib.pyplot as plt

# =========================
# 1. 함수 정의
# =========================
x = np.linspace(-3, 3, 80)
y = np.linspace(-3, 3, 80)

X, Y = np.meshgrid(x, y)

# 3차원 이차함수
Z = X**2 + Y**2

# =========================
# 2. Figure 생성
# =========================
fig = plt.figure(figsize=(7, 5), facecolor="none")
ax = fig.add_subplot(111, projection="3d")

# =========================
# 3. 색 없는 3D 그리드 그래프
# =========================
ax.plot_wireframe(
    X, Y, Z,
    rstride=4,
    cstride=4,
    linewidth=0.8,
    color = "black"
)

# =========================
# 4. 축, 눈금, 배경 제거
# =========================
ax.set_axis_off()
ax.grid(False)

fig.patch.set_alpha(0)
ax.patch.set_alpha(0)

# =========================
# 5. 시점 조절
# =========================
ax.view_init(elev=28, azim=-55)

# 여백 제거
plt.subplots_adjust(left=0, right=1, bottom=0, top=1)

# =========================
# 6. 투명 PNG로 저장
# =========================
plt.savefig(
    "transparent_3d_quadratic_wireframe.png",
    dpi=300,
    transparent=True,
    bbox_inches="tight",
    pad_inches=0
)

plt.show()
```
